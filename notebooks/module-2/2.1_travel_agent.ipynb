# Travel Agent
#### Trying out the [Kiwi](https://apify.com/agentify/kiwi-mcp-server) online MCP server for fetching **flight details**.

```python
# Base utils.
from IPython.display import Markdown

# Model init. and invocation
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
from langchain.agents import create_agent

# MCP server
from langchain_mcp_adapters.client import MultiServerMCPClient
```

```python
client = MultiServerMCPClient(
    {
        "travel_server": {
                "transport": "streamable_http",
                "url": "https://mcp.kiwi.com"
            }
    }
)

tools = await client.get_tools()
```

```python
model = "ollama:gemma4:31b-cloud"

agent = create_agent(
    model=model,
    tools=tools,
    checkpointer=InMemorySaver(),
    system_prompt="You are a travel agent. Do not ask any follow up questions. Also, do not provide any unnecessary details, like recommendations or facts. Just provide the travel plan details (if possible) in tabular formmat and keep it short."
)
```

```python
config = {"configurable": {"thread_id": "1"}}

response = await agent.ainvoke(
    {"messages": [HumanMessage(content="Fetch me a direct flight details from New Delhi to Lucknow on August 4th, 2026.")]},
    config # type: ignore
)
```
